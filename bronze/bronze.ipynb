```python
bronze_df=spark.read.format('csv').option('header','true').option('inferSchema','true').load('/Volumes/wfrgbhn/bronze/raw_sales_volume/sales_order_raw.csv')

bronze_df.show()
```

```python
from pyspark.sql.functions import col, current_timestamp

bronze_sales_df=bronze_df.withColumn("source_file",col("_metadata.file_path")).withColumn("ingestion_timestamp", current_timestamp())
```

```python
bronze_sales_df = bronze_sales_df.withColumnRenamed("unit_price ", "unit_price")
bronze_sales_df.write.format('delta').mode('overwrite').saveAsTable('wfrgbhn.bronze.bronze_sales_delta')
```
